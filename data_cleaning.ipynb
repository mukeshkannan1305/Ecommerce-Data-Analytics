```python
df = pd.read_csv("../data/raw/sales.csv")
```

```python
df = df.drop_duplicates()
```

```python
df[df["Sales"].isnull()]
```

```python
df.loc[df["Sales"].isnull(), "Sales"] = (
    df.loc[df["Sales"].isnull(), "Profit"] +
    df.loc[df["Sales"].isnull(), "Cost"]
) / (1 - df.loc[df["Sales"].isnull(), "Discount"])
```

```python
df["Sales"].isnull().sum()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df["Order_Date"].dtype
```

```python
#####tjhis after
```

```python
df.describe()
```

```python
df["Calculated_Profit"] = df["Sales"] - df["Cost"]
```

```python
(df["Profit"] - df["Calculated_Profit"]).describe()
```

```python
df["Calculated_Profit"] = df["Sales"] - df["Cost"] - (df["Sales"] * df["Discount"])
```

```python
(df["Profit"] - df["Calculated_Profit"]).describe()
```

```python
df.drop("Calculated_Profit", axis=1, inplace=True)
```

```python
df["Year"] = df["Order_Date"].dt.year
```

```python
df[["Order_Date", "Year"]].head()
```

```python
df["Month"] = df["Order_Date"].dt.month
```

```python
df[["Order_Date", "Year", "Month"]].head()
```

```python
df["Month_Name"] = df["Order_Date"].dt.month_name()
```

```python
df[["Order_Date", "Year", "Month", "Month_Name"]].head()
```

```python
df["Quarter"] = df["Order_Date"].dt.quarter
```

```python
df[["Order_Date", "Year", "Month", "Month_Name", "Quarter"]].head()
```

```python
df["Quarter_Name"] = "Q" + df["Quarter"].astype(str)
```

```python
df[["Order_Date", "Year", "Month_Name", "Quarter", "Quarter_Name"]].head()
```

```python
df["Discount_Amount"] = df["Sales"] * df["Discount"]
```

```python
df[["Sales", "Discount", "Discount_Amount"]].head()
```

```python
df["Profit_Margin"] = (df["Profit"] / df["Sales"]) * 100

df[["Sales", "Profit", "Profit_Margin"]].head()
```

```python
df["Return_Flag_Num"] = df["Return_Flag"].map({
    "Yes": 1,
    "No": 0
})

df[["Return_Flag", "Return_Flag_Num"]].head()
```

```python
df.to_csv("sales_data_cleaned.csv", index=False)
```
